## Note for Further Runs: 
If the delta tables already exist and there are errors occuring with the writing of the stream df from this ETL process being ran again the tables need to be deleted in the `Catalog/Legacy/hive_metastore/default/table_name` path. 

Then you need to run the following code cell to with the corresponding path, E.g.:
`"/tmp/kinesis/_checkpoints/pin/"`

As defined in the `write_stream_df_to_delta_table()` function. 

This will delete the stream Checkpoint which is not preferred as it allows Spark to resume the stream exactly where it left off, but some of the errors found in this project may deem it necessary to do so.

```python
stream_name = 'pin'
dbutils.fs.rm(f"/tmp/kinesis/_checkpoints/{stream_name}", True)
```
